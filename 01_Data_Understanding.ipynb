# HOSPITAL PATIENT FLOW & DEMAND PREDICTION SYSTEM
## NOTEBOOK 01: DATA UNDERSTANDING
**Workflow Phase 01** of the 17-Stage Hospital Operations Decision Support Architecture.

### Objectives:
1. Ingest raw hospital operational records (`admissions.csv`, 120,000 records).
2. Inspect dataset structure, feature names, and data types.
3. Validate missing values, duplicated entries, and data cleanliness.
4. Verify chronological date spans across the 10-year observational horizon (2015 - 2024).
5. Examine categorical distributions across wards, admission modes, and discharge categories.


```python
# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import os

print("Libraries imported successfully.")
```

```python
# 2. LOAD HOSPITAL ADMISSIONS DATASET
dataset_path = 'admissions.csv'
df = pd.read_csv(dataset_path)

print(f"Dataset successfully loaded.")
print(f"Total Records: {len(df):,}")
print(f"Total Features: {df.shape[1]}")
df.head(5)
```

```python
# 3. DATA TYPES & NON-NULL SUMMARY
df.info()
```

```python
# 4. MISSING VALUES & DUPLICATES CHECK
print("=== MISSING VALUES PER FEATURE ===")
print(df.isnull().sum())
print(f"\nTotal Missing Cells: {df.isnull().sum().sum()}")
print(f"Duplicate Rows Count: {df.duplicated().sum()}")
if 'admission_id' in df.columns:
    print(f"Duplicate Admission IDs: {df['admission_id'].duplicated().sum()}")
```

```python
# 5. CHRONOLOGICAL DATE SPAN VERIFICATION
admit_dates = pd.to_datetime(df['admit_date'])
discharge_dates = pd.to_datetime(df['discharge_date'])

print(f"Earliest Admission Date:  {admit_dates.min().strftime('%Y-%m-%d')}")
print(f"Latest Admission Date:    {admit_dates.max().strftime('%Y-%m-%d')}")
print(f"Earliest Discharge Date:  {discharge_dates.min().strftime('%Y-%m-%d')}")
print(f"Latest Discharge Date:    {discharge_dates.max().strftime('%Y-%m-%d')}")
print(f"Total Span:               {(admit_dates.max() - admit_dates.min()).days:,} days (approx. 10 years)")
```

```python
# 6. WARD & ADMISSION TYPE DISTRIBUTIONS
print("=== WARD DISTRIBUTION ===")
print(df['ward_type'].value_counts(normalize=True).round(4) * 100)
print("\n=== ADMISSION TYPE DISTRIBUTION ===")
print(df['admit_type'].value_counts(normalize=True).round(4) * 100)
print("\n=== DISCHARGE TYPE DISTRIBUTION ===")
print(df['discharge_type'].value_counts())
```

```python
# 7. CLINICAL & NUMERICAL SUMMARY STATISTICS
numerical_cols = ['los_days', 'num_procedures', 'charlson_index', 'hba1c', 'creatinine', 'haemoglobin', 'systolic_bp']
df[numerical_cols].describe().T
```
